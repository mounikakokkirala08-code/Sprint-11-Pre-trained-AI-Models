## Architecture

Architecture means the internal structure/design of a model — how its layers are arranged and how data flows through them.

| Factor             | ResNet18                           | MobileNetV2                                                           |
| ------------------ | ---------------------------------- | --------------------------------------------------------------------- |
| **Architecture**   | CNN with residual/skip connections | CNN using depthwise separable convolutions + inverted residual blocks |
| **Parameters**     | ~11.7M                             | ~3.5M                                                                 |
| **Model Size**     | Larger                             | Smaller                                                               |
| **Inference Time** | Generally higher                   | Generally lower                                                       |
| **Memory Usage**   | Higher                             | Lower                                                                 |
| **Output Quality** | High                               | Good                                                                  |
| **Best For**       | Accuracy-focused tasks             | Lightweight/mobile tasks                                              |


Architecture describes the internal design of a model, including its layers, connections, and operations. Different architectures affect parameters, inference speed, memory usage, and output quality.

| Factor                     | ResNet18                                                 | MobileNetV2                                                 |
| -------------------------- | -------------------------------------------------------- | ----------------------------------------------------------- |
| **Output**                 | Class label + confidence score                           | Class label + confidence score                              |
| **Accuracy / Performance** | Generally high                                           | Generally good                                              |
| **Inference Speed**        | Medium                                                   | Fast                                                        |
| **Hardware Requirements**  | CPU works; GPU recommended for faster processing         | CPU-friendly; GPU optional                                  |
| **License**                | BSD-3-Clause                                             | Apache-2.0                                                  |
| **Limitations**            | Larger and more resource-intensive                       | May sacrifice some accuracy for efficiency                  |
| **Use Cases**              | Image classification, computer vision, transfer learning | Mobile apps, edge devices, lightweight image classification |
