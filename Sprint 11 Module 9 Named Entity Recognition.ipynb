## Named Entity Recognition

## Step 1 – Explain the Concept

NER identifies important entities in text and gives each entity a category/label.

Common labels:

PER → Person

ORG → Organization

LOC → Location

DATE → Date

MONEY → Money

## Step 2 – Simple Example

**“Named Entity Recognition is an NLP task that identifies and classifies entities such as persons, organizations, and locations from text.”**

What is NER?

NER (Named Entity Recognition) is an NLP technique that finds important names/entities in text and identifies what type they are.

Simple Example

Sentence:

“Mounika works at Cognizant in Hyderabad.”

NER identifies:

| Entity    | Type         |
| --------- | ------------ |
| Mounika   | Person       |
| Cognizant | Organization |
| Hyderabad | Location     |


**“NER is an NLP task that identifies and classifies named entities such as persons, organizations, locations, dates, and money from text.”**

NER – Entity Types

Entity types are the categories that NER assigns to identified words or phrases.

| Entity Type            | Meaning              | Example           |
| ---------------------- | -------------------- | ----------------- |
| **PERSON (PER)**       | Person's name        | Mounika           |
| **ORGANIZATION (ORG)** | Company/organization | Cognizant         |
| **LOCATION (LOC)**     | Location/place       | Hyderabad         |
| **GPE**                | Geopolitical entity  | India             |
| **DATE**               | Date                 | 5 October 2026    |
| **TIME**               | Time                 | 10:30 AM          |
| **MONEY**              | Monetary value       | ₹50,000           |
| **PERCENT**            | Percentage           | 20%               |
| **PRODUCT**            | Product              | iPhone            |
| **EVENT**              | Event                | Olympics          |
| **FACILITY**           | Building/place       | Hyderabad Airport |


## Token-Level Prediction in NER
## Step 1 – Explain the Concept

Token-level prediction means the NER model predicts an entity label for each token in the input text.

A token can be a word or part of a word.

## Step 2 – Simple Example

Sentence:

“Mounika works at Cognizant.”

### Step 3 – BIO Format

NER commonly uses BIO labels:

B = Beginning of entity
I = Inside entity
O = Outside entity

## Entity Grouping in NER
### Step 1 – Explain the Concept

Entity grouping means combining multiple tokens that belong to the same entity into one complete entity.

### Step 2 – Simple Example

Sentence:

“Mounika Kokkirala works at Cognizant Technology Solutions.”

Token-level predictions:

** “NER identifies entities in text, predicts labels at the token level, and groups related tokens to produce complete entities such as persons, organizations, locations, and dates.”
