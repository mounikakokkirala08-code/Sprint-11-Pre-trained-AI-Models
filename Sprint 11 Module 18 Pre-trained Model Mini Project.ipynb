```python
from transformers import pipeline

classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)
```

```python
text = "Artificial intelligence and machine learning are transforming software."

result = classifier(
    text,
    candidate_labels=["Technology", "Sports", "Business"]
)

print(result)
```

```python
prediction = result["labels"][0]
score = result["scores"][0]

print("Category:", prediction)
print("Confidence:", score)
```

```python
documents = [
    "AI models are improving computer vision.",
    "The player scored two goals in the final match.",
    "The company increased its revenue this quarter."
]

labels = ["Technology", "Sports", "Business"]

for text in documents:
    result = classifier(text, candidate_labels=labels)

    print("Document:", text)
    print("Prediction:", result["labels"][0])
    print("Score:", round(result["scores"][0], 3))
    print()
```
