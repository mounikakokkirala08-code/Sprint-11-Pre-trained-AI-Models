## Introduction to Pre-trained Models

A pre-trained model is a machine learning model that has already been trained on a large dataset.

Instead of training a model from the beginning, we can reuse the learned knowledge for our own task.

## What is a Model?
### Step 1 – Explain the Concept

A model is a system that learns patterns from data and uses those patterns to make predictions or decisions.

In simple words:

Model = A trained system that learns from data and gives an output.

### Step 2 – Simple Example / Real-world Analogy

Suppose we give a model information about houses:

Area → 1000 sq.ft
Bedrooms → 2
Location → Hyderabad

The model learns from previous house data and predicts:

Predicted Price → ₹50 Lakhs

So:

Data → Model → Prediction

### Step 3 – Implement

Simple Python example:

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

X = [[1000], [1500], [2000]]
y = [30, 45, 60]

model.fit(X, y)

prediction = model.predict([[1800]])

print(prediction)
```

Training from Scratch vs Pre-trained Models
### Step 1 – Concept
Training from scratch: Train a model from zero using your own data.
Pre-trained model: Start with a model already trained on a large dataset.
### Step 2 – Simple Example

👨‍🎓 From scratch: Learn a subject from ABC level.
🎓 Pre-trained: Already know the basics and learn only the new topic.

### Step 4 – Explain

|                | From Scratch     | Pre-trained               |
| -------------- | ---------------- | ------------------------- |
| Starting point | Random weights   | Learned weights           |
| Data needed    | More             | Usually less              |
| Training time  | More             | Less                      |
| Cost           | Higher           | Lower                     |
| Knowledge      | Learns from zero | Reuses existing knowledge |


## Model Weights vs Model Parameters
### Step 1 – Concept
Model Parameters: Values the model learns during training.
Model Weights: Numerical values that determine how strongly one input affects the output.

Weights are one type of model parameter.

### Step 2 – Simple Example

Imagine:

Study Hours → Exam Score

If the model learns:

Weight = 10

It means study hours have a strong effect on the predicted score.

### Step 3 – Implement
  # Bias
### Step 4 – Explain
Input: Study hours
Weight: Learned relationship between input and output
Bias: Starting/base value
Output: Prediction
During training, the model updates weights and biases.

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

X = [[1], [2], [3]]
y = [10, 20, 30]

model.fit(X, y)

print(model.coef_)       # Weight
print(model.intercept_)
```

## Checkpoints
### Step 1 – Concept

A checkpoint is a saved copy of a model during training.

Checkpoint = Saved model state at a particular training point.

### Step 2 – Simple Example

Like saving a game after reaching Level 5.
If something goes wrong, you can continue from Level 5 instead of starting again.

### Step 3 – Implement
import torch

torch.save(model.state_dict(), "checkpoint.pth")

### Step 4 – Explain
Input: Trained model
Action: Save model weights
Output: checkpoint.pth
Why: To resume training or reuse the model later.
### Step 5 – Experiment
torch.save(model.state_dict(), "checkpoint_epoch10.pth")

Now the checkpoint is saved after epoch 10.

### Pre-trained Model Workflow
### Step 1 – Explain the Concept

The workflow is:

Dataset
   ↓
Training
   ↓
Model Weights
   ↓
Pre-trained Model
   ↓
Inference
Dataset → Data used for learning.
Training → Model learns patterns from the data.
Model Weights → Learned values stored by the model.
Pre-trained Model → Model with already learned weights.
Inference → Use the trained model to make predictions.
### Step 2 – Simple Example

For fruit classification:

Apple + Banana + Orange Images
          ↓
       Training
          ↓
   Learned Weights
          ↓
   Pre-trained Model
          ↓
    New Fruit Image
          ↓
      Prediction

Example output:

Prediction: Apple
