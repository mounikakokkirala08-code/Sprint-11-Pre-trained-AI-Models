## Speech-to-Text (STT)
### Step 1 – Explain the Concept

Speech-to-Text means converting spoken audio into written text.

It is also called Automatic Speech Recognition (ASR).

### Step 2 – Simple Example

🎤 Input:

“I am learning Python.”

⬇️ Speech Recognition Model

📝 Output:

"I am learning Python."

### Step 3 – Implement

Using an open-source pre-trained Whisper model:

```python
from transformers import pipeline

stt = pipeline(
    "automatic-speech-recognition",
    model="openai/whisper-small"
)

result = stt("left audio.wav")

print(result["text"])
```

```python
## Sampling Rate
### Step 1 – Explain the Concept

Sampling rate is the number of audio samples captured per second.

It is measured in Hertz (Hz).

Example:

16,000 Hz (16 kHz) → 16,000 samples are captured every second.
Higher sampling rate → captures more audio detail.
### Step 2 – Simple Example

Think of taking photos of a moving object 📸:

10 photos/second → less detail about movement
100 photos/second → more detail

Similarly, audio is measured many times per second.

Audio → Samples → Digital Audio

### Step 3 – Python Example

import librosa

audio, sample_rate = librosa.load("audio.wav", sr=16000)

print("Sampling Rate:", sample_rate)
```

## Audio Preprocessing
## Step 1 – Explain the Concept

Audio preprocessing means converting raw audio into a clean format that the speech model can understand easily.

Common steps:

Raw Audio → Resampling → Noise Reduction → Normalization → Features → Model

## Step 2 – Simple Example

Suppose you record:

🎤 “I am learning Python”

The audio may have:

Background noise
Different sampling rate
Different volume
Stereo/mono differences

Preprocessing prepares it before giving it to the model.

```python
from transformers import pipeline

# Load pre-trained speech-to-text model
stt = pipeline(
    "automatic-speech-recognition",
    model="openai/whisper-small"
)

# Audio Sample 1
result1 = stt("left audio.wav")
print("Audio 1:", result1["text"])

# Audio Sample 2
result2 = stt("right audio.wav")
print("Audio 2:", result2["text"])
```
