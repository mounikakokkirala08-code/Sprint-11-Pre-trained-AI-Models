| **Problem**                 | What problem do we want to solve?       | Classify documents into categories                            |
| **Input Type**              | What data will we give the model?       | Text/document                                                 |
| **Expected Output**         | What result do we need?                 | Category: Sports, Business, Technology                        |
| **Suitable Model Type**     | What type of model is appropriate?      | Text classification model                                     |
| **Candidate Models**        | Possible models we can choose from.     | BERT, RoBERTa, BART                                           |
| **Selection Criteria**      | Factors used to choose the best model.  | Accuracy, task support, size, license                         |
| **Hardware Considerations** | Hardware needed to run the model.       | CPU/GPU, RAM, storage                                         |
| **Limitations**             | Problems or weaknesses of the model.    | May struggle with ambiguous documents                         |
| **Final Justification**     | Explain why the selected model is best. | Good task support + suitable performance + available hardware |
