## Tokenization

### 1. What is Tokenization?

Tokenization is the process of breaking text into smaller pieces called tokens so that a model can process the text.

Example:

Input:
"I love machine learning"

        ↓ Tokenization

["I", "love", "machine", "learning"]

The tokens are then converted into numbers (token IDs) because the model works with numerical input.

Text
 ↓
Tokens
 ↓
Token IDs
 ↓
Model
### 2. Why is Tokenization needed?

AI models cannot directly understand raw text like humans.

They need numerical representations.

"I love AI"

    ↓
["I", "love", "AI"]

    ↓
[40, 1567, 890]

    ↓
Model

### 3. Types of Tokenization

| Type            | Example                               |
| --------------- | ------------------------------------- |
| Word-level      | `"I love AI"` → `["I", "love", "AI"]` |
| Character-level | `"cat"` → `["c", "a", "t"]`           |
| Subword-level   | `"playing"` → `["play", "##ing"]`     |


```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

text = 'Machine learninng is branch of artifical inteligence'

token = tokenizer.tokenize(text)

print(token)
```

“Tokenization is the process of converting text into smaller units called tokens. These tokens are then converted into numerical token IDs that a Transformer model can understand. For example, ‘I love AI’ can be split into tokens such as ‘I’, ‘love’, and ‘AI’, and each token is mapped to an ID. Hugging Face provides tokenizers to perform this preprocessing automatically.”

### What is a Token?

A token is a small piece of text that a language model processes.

A token can be:

a word
part of a word
a character
punctuation
Example
"I love AI!"
       ↓
["I", "love", "AI", "!"]

“A token is a small unit of text processed by a language model. It can be a word, part of a word, or punctuation. During tokenization, the input text is split into tokens, and each token is converted into a numerical token ID that the model can process.”


### Why is Tokenization Required?
Simple answer;  Tokenization is required because language models cannot directly process raw text. They need text to be converted into smaller units called tokens and then into numerical token IDs.

| Type                   | Meaning                                     | Example                               |
| ---------------------- | ------------------------------------------- | ------------------------------------- |
| **1. Word-level**      | Splits text into complete words             | `"I love AI"` → `["I", "love", "AI"]` |
| **2. Character-level** | Splits text into individual characters      | `"cat"` → `["c", "a", "t"]`           |
| **3. Subword-level**   | Splits words into smaller meaningful pieces | `"playing"` → `["play", "##ing"]`     |


“There are three common types of tokenization: word-level, character-level, and subword-level tokenization. Word-level splits text into complete words, character-level splits it into individual characters, and subword-level splits words into smaller pieces. Modern Transformer models commonly use subword tokenization because it handles unknown and rare words more effectively.”


```python
### What are Token IDs?

Token ID is a unique number assigned to each token in a tokenizer's vocabulary.

The model works with numbers, not directly with text. So after tokenization, each token is converted into a token ID.

Simple Example
Text
"I love AI"
   ↓
Tokens
["I", "love", "AI"]
   ↓
Token IDs
[40, 2293, 993]

Important: These numbers are only an example. Actual token IDs depend on the tokenizer/model.

Complete Flow
"I love AI"
     ↓
Tokenization
     ↓
["I", "love", "AI"]
     ↓
Token IDs
[40, 2293, 993]
     ↓
Embeddings
     ↓
Transformer Model
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

text = "i love AI"

tokens = tokenizer.tokenize(text)

token_id = tokenizer.convert_tokens_to_ids(tokens)

print(token_id)
```

“Token IDs are numerical IDs assigned to tokens by a tokenizer. Since neural networks cannot directly process text, the tokens are converted into numbers. These token IDs are then converted into embeddings and passed to the Transformer model.”


## Vocabulary

In NLP, vocabulary is the complete collection of tokens that a tokenizer knows.

Think of it like a dictionary.

Simple Example

Suppose the vocabulary contains:

"I"      → 101
"love"   → 202
"AI"     → 303
"Python" → 404

Here:

Token = "love"
Token ID = 202
Vocabulary = {"I", "love", "AI", "Python", ...}

## Special Tokens

Special tokens are predefined tokens used by the model to understand special information or control how the input/output should be processed.

They are different from normal words like "I", "love", or "AI".

| Special Token | Meaning                                           |
| ------------- | ------------------------------------------------- |
| `[CLS]`       | Start/representation token, commonly used in BERT |
| `[SEP]`       | Separates sentences or segments                   |
| `[PAD]`       | Padding shorter sequences to the same length      |
| `[MASK]`      | Hides a token for masked-token prediction         |
| `[UNK]`       | Unknown token                                     |
| `<BOS>`       | Beginning of sequence                             |
| `<EOS>`       | End of sequence                                   |


## Attention Mask

An attention mask tells the Transformer which tokens it should pay attention to and which tokens it should ignore.

Simple Example

Suppose we have two sentences of different lengths:

Sentence 1: I love AI

Sentence 2: I love AI very much

To process them together, we may add padding:

I love AI [PAD] [PAD]

I love AI very much

The attention mask tells the model:

I    love   AI   PAD   PAD

1     1     1     0     0

1 → token should be considered

0 → token should be ignored

Why is Attention Mask Needed?

Mainly for padding.

Real tokens → 1
Padding     → 0

This prevents the model from treating [PAD] as meaningful information.

Hugging Face Example

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

text = "I love india country its my country its my native place"

result = tokenizer(text, padding = "max_length", max_length = 6)

print(result)
```

“An attention mask tells the Transformer which tokens should be considered during attention and which tokens should be ignored. A common use is masking padding tokens. Typically, 1 indicates a valid token and 0 indicates a token that should be ignored. In decoder models, causal masking is used separately to prevent the model from looking at future tokens.”


```python
Attention Mask → usually ignores padding.
Causal Mask → prevents looking at future tokens.
```

## Padding and Truncation

These are tokenization techniques used to make input sequences the required length for a model.

1. Padding

Padding means adding special [PAD] tokens to shorter sequences so all sequences have the same length.

Example: Suppose max_length = 5

"I love AI"
       ↓
[I, love, AI, PAD, PAD]

The attention mask tells the model to ignore those padding tokens:

Tokens:          I   love   AI   PAD   PAD
Attention Mask:  1    1     1     0     0

## 2. Truncation

Truncation means removing tokens from a sequence when it is longer than the required maximum length.

Example: max_length = 5

"I love machine learning and artificial intelligence"
                     ↓
             Keep only 5 tokens

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

text = "I love machine learning"

result = tokenizer(
    text,
    padding="max_length",
    truncation=True,
    max_length=8
)

print(result["input_ids"])
print(result["attention_mask"])
```

**“Padding means adding special padding tokens to shorter sequences so that all inputs have the same length. Truncation means removing extra tokens from sequences that are longer than the maximum allowed length. In Hugging Face, we can control both using padding, truncation, and max_length parameters.”**

**“Maximum sequence length is the maximum number of tokens that a Transformer model can process in a single input sequence. If the input is shorter, we can use padding, and if it is longer, we use truncation. The maximum length depends on the model architecture and its supported context length.”**


```python
from transformers import AutoTokenizer

# Load BERT tokenizer
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# Different sentences
sentences = [
    "I love AI.",
    "Python is easy to learn.",
    "Transformers are powerful."
]

for text in sentences:

    print("=" * 60)

    # Original Text
    print("Original Text:")
    print(text)

    # Tokenization
    tokens = tokenizer.tokenize(text)
    print("\nTokens:")
    print(tokens)

    # Token IDs
    token_ids = tokenizer.convert_tokens_to_ids(tokens)
    print("\nToken IDs:")
    print(token_ids)

    # Attention Mask
    result = tokenizer(
        text,
        padding="max_length",
        truncation=True,
        max_length=10
    )

    print("\nAttention Mask:")
    print(result["attention_mask"])
```
