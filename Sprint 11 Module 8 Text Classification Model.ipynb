## Text Classification Model

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

# model name

model_name = "distilbert-base-uncased-finetuned-sst-2-english"
```

```python
# 2. Load Tokenizer

tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
# 3. Load Pre-trained Classification Model
# -----------------------------------------
# This model is already trained for sentiment classification
model = AutoModelForSequenceClassification.from_pretrained(model_name)
```

```python
# 4. Give input text
# -----------------------------------------
text = "I really enjoyed this movie!"
```

```python
# 5. Tokenize the text
# -----------------------------------------
inputs = tokenizer(
    text,
    return_tensors="pt"
)

print("Model Inputs:")
print(inputs)
```

```python
#6. Send inputs to the model
# -----------------------------------------
outputs = model(**inputs)
```

```python
# -----------------------------------------
# 7. Get raw model output
# -----------------------------------------
logits = outputs.logits

print("\nLogits:")
print(logits)
```

```python
# -----------------------------------------
# 8. Convert logits into probabilities
# -----------------------------------------
probabilities = torch.softmax(logits, dim=-1)

print("\nProbabilities:")
print(probabilities)


# -----------------------------------------
# 9. Find the highest probability
# -----------------------------------------
prediction = torch.argmax(probabilities, dim=-1)

print("\nPrediction ID:")
print(prediction.item())


# -----------------------------------------
# 10. Convert ID into label
# -----------------------------------------
labels = ["NEGATIVE", "POSITIVE"]

print("\nFinal Prediction:")
print(labels[prediction.item()])
```

I used an open-source pre-trained DistilBERT model fine-tuned for sentiment classification. First, I loaded the tokenizer and classification model using Hugging Face `AutoTokenizer` and `AutoModelForSequenceClassification`.

I tested multiple inputs including positive, negative, neutral, and ambiguous sentences. The tokenizer converted the text into `input_ids` and `attention_mask`, which were passed to the model during inference. The model produced raw logits, which I converted into probabilities using Softmax. I then selected the class with the highest probability and displayed the prediction and confidence score.

During analysis, I observed that neutral and ambiguous inputs can produce unexpected predictions because this particular model has only two classes: POSITIVE and NEGATIVE. Therefore, the model is forced to select one of those classes even when the input does not clearly express positive or negative sentiment.


```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch


# =========================================================
# 1. SELECT A PRE-TRAINED MODEL
# =========================================================

model_name = "distilbert-base-uncased-finetuned-sst-2-english"


# =========================================================
# 2. LOAD TOKENIZER
# =========================================================
# Tokenizer converts text into tokens and token IDs.

tokenizer = AutoTokenizer.from_pretrained(model_name)


# =========================================================
# 3. LOAD MODEL
# =========================================================
# This model is already trained for sentiment classification.
# It predicts:
# 0 -> NEGATIVE
# 1 -> POSITIVE

model = AutoModelForSequenceClassification.from_pretrained(model_name)


# Labels used by the model
labels = ["NEGATIVE", "POSITIVE"]


# =========================================================
# 4. CREATE MULTIPLE TEST INPUTS
# =========================================================

texts = {
    "Positive": "I really enjoyed this movie!",
    
    "Negative": "This movie was terrible and boring.",
    
    "Neutral": "The movie starts at 7 PM.",
    
    "Ambiguous": "The movie was okay, I guess."
}


# =========================================================
# 5. TEST EACH INPUT
# =========================================================

for category, text in texts.items():

    print("\n" + "=" * 70)

    # -----------------------------------------------------
    # Original input
    # -----------------------------------------------------
    print("Test Type:", category)
    print("Original Text:", text)


    # -----------------------------------------------------
    # 6. TOKENIZE
    # -----------------------------------------------------
    # Converts text into input_ids and attention_mask.

    inputs = tokenizer(
        text,
        return_tensors="pt"
    )

    print("\nInput IDs:")
    print(inputs["input_ids"])

    print("\nAttention Mask:")
    print(inputs["attention_mask"])


    # -----------------------------------------------------
    # 7. RUN INFERENCE
    # -----------------------------------------------------
    # torch.no_grad() means we are only predicting.
    # We are NOT training the model.

    with torch.no_grad():
        outputs = model(**inputs)


    # -----------------------------------------------------
    # 8. GET MODEL OUTPUT
    # -----------------------------------------------------
    # Logits are raw scores produced by the model.

    logits = outputs.logits

    print("\nLogits:")
    print(logits)


    # -----------------------------------------------------
    # 9. PROCESS OUTPUT
    # -----------------------------------------------------
    # Softmax converts logits into probabilities.

    probabilities = torch.softmax(logits, dim=-1)

    print("\nProbabilities:")
    print(probabilities)


    # -----------------------------------------------------
    # 10. FIND HIGHEST SCORE
    # -----------------------------------------------------

    prediction_id = torch.argmax(probabilities, dim=-1).item()

    prediction = labels[prediction_id]

    confidence = probabilities[0][prediction_id].item()


    # -----------------------------------------------------
    # 11. DISPLAY PREDICTION
    # -----------------------------------------------------

    print("\nPrediction:", prediction)

    print("Confidence:", round(confidence * 100, 2), "%")


    # -----------------------------------------------------
    # 12. DISPLAY BOTH SCORES
    # -----------------------------------------------------

    print("\nScore Details:")

    print(
        "NEGATIVE:",
        round(probabilities[0][0].item() * 100, 2),
        "%"
    )

    print(
        "POSITIVE:",
        round(probabilities[0][1].item() * 100, 2),
        "%"
    )
```
