```python
InferenceService
Step 1: Simple Definition

InferenceService is a component that takes new input data and sends it to a trained/pre-trained model to get a prediction.

In simple words:
```

```python
class InferenceService:

    def __init__(self, loader):
        self.loader = loader

    def predict_sentiment(self, text):
        return self.loader.sentiment_model(text)

    def predict_entities(self, text):
        return self.loader.ner_model(text)
```

```python
Preprocessor
Step 1: Simple Definition

A Preprocessor is a component that cleans and prepares input data before sending it to the AI model.
```

```python
class Preprocessor:

    def clean_text(self, text):

        # Remove extra spaces
        text = text.strip()

        # Convert multiple spaces into one
        text = " ".join(text.split())

        return text
```

```python
PostProcessor
Step 1: Simple Definition

A PostProcessor is a component that processes the model's raw prediction and converts it into a clean, useful result.

In simple words:

PostProcessor = Clean and organize the model output after prediction.
```

```python
class PostProcessor:

    def process_sentiment(self, result):

        return {
            "label": result[0]["label"],
            "confidence": round(result[0]["score"], 4)
        }
```

```python
Separation of Responsibilities
Step 1: Simple Definition

Separation of Responsibilities means:

Each component should have one specific job.

Instead of putting all code in one file, we divide the application into different components.

For your AI pipeline:

PreProcessor       → Prepare input
ModelConfig        → Store model settings
ModelLoader        → Load models
InferenceService   → Get predictions
PostProcessor      → Process output
Formatter          → Format result
Streamlit          → Show result

This makes the project clean, easy to understand, test, a
```
