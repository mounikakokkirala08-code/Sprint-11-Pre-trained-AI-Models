## Loading Models Manually in Hugging Face

Loading a model manually means instead of using the simple pipeline(), we directly load the tokenizer and model using AutoTokenizer and AutoModel....

Simple difference

| Approach       | Example                       | Use                         |
| -------------- | ----------------------------- | --------------------------- |
| `pipeline()`   | High-level API                | Easy, quick inference       |
| Manual loading | `AutoTokenizer` + `AutoModel` | More control over the model |


```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

# Model name from Hugging Face
model_name = "distilbert-base-uncased-finetuned-sst-2-english"

# 1. Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_name)

# 2. Load pre-trained model
model = AutoModelForSequenceClassification.from_pretrained(model_name)

# 3. Input text
text = "I love this product!"

# 4. Tokenize
inputs = tokenizer(
    text,
    return_tensors="pt"
)

# 5. Send input to model
outputs = model(**inputs)

# 6. Get logits
logits = outputs.logits

print(logits)
```

In Hugging Face, loading a model manually means directly loading the tokenizer and pre-trained model instead of using the high-level `pipeline()` API.

We commonly use `AutoTokenizer.from_pretrained()` to load the tokenizer and `AutoModel` or a task-specific class such as `AutoModelForSequenceClassification.from_pretrained()` to load the pre-trained model and its learned weights.

The tokenizer converts text into inputs such as `input_ids` and `attention_mask`, and the model processes these inputs to produce outputs such as logits or predictions.

The basic flow is:

**Text → Tokenizer → Token IDs + Attention Mask → Model → Output**

Manual loading gives us more control over tokenization, model inputs, outputs, and inference compared with a pipeline.


| Pipeline                               | Manual                               |
| -------------------------------------- | ------------------------------------ |
| `pipeline()` does things automatically | We do things step-by-step            |
| Easy                                   | More control                         |
| Less code                              | More code                            |
| Good for beginners                     | Good for understanding/customization |


Manual model loading means loading the tokenizer and the pre-trained model separately using Hugging Face.

The tokenizer converts text into token IDs, and the model processes those token IDs to produce the output.

We commonly use `AutoTokenizer.from_pretrained()` for the tokenizer and `AutoModel.from_pretrained()` or a task-specific model class for the model.

**Simple flow:**

Text → Tokenizer → Token IDs → Model → Output


```python
## Auto Loading using pipeline()

# Here, Hugging Face handles the tokenizer and model for us.

from transformers import pipeline

# Create a sentiment analysis pipeline
# Hugging Face automatically loads:
# 1. Tokenizer
# 2. Pre-trained model
# 3. Prediction process

classifier = pipeline("sentiment-analysis")

# Give text to the pipeline
text = "I love this movie!"

# Get prediction
result = classifier(text)

print(result)
```

```python
## Manual load 

from transformers import AutoTokenizer
from transformers import AutoModelForSequenceClassification
import torch

# Name of the pre-trained model
model_name = "distilbert-base-uncased-finetuned-sst-2-english"

# ------------------------------------------------
# STEP 1: Load the tokenizer
# ------------------------------------------------
# Tokenizer converts text into token IDs
tokenizer = AutoTokenizer.from_pretrained(model_name)


# ------------------------------------------------
# STEP 2: Load the pre-trained model
# ------------------------------------------------
# Model contains the already-trained weights
model = AutoModelForSequenceClassification.from_pretrained(model_name)


# ------------------------------------------------
# STEP 3: Give input text
# ------------------------------------------------
text = "I love this movie!"


# ------------------------------------------------
# STEP 4: Convert text into model inputs
# ------------------------------------------------
# return_tensors="pt" means return PyTorch tensors
inputs = tokenizer(
    text,
    return_tensors="pt"
)

print("Model Inputs:")
print(inputs)


# ------------------------------------------------
# STEP 5: Give inputs to the model
# ------------------------------------------------
# The model produces raw scores called logits
outputs = model(**inputs)

print("Logits:")
print(outputs.logits)


# ------------------------------------------------
# STEP 6: Convert logits into probabilities
# ------------------------------------------------
probabilities = torch.softmax(outputs.logits, dim=-1)

print("Probabilities:")
print(probabilities)


# ------------------------------------------------
# STEP 7: Find the class with highest probability
# ------------------------------------------------
prediction = torch.argmax(probabilities, dim=-1)

print("Predicted class ID:")
print(prediction.item())


# ------------------------------------------------
# STEP 8: Convert class ID into label
# ------------------------------------------------
labels = ["NEGATIVE", "POSITIVE"]

print("Prediction:")
print(labels[prediction.item()])
```
