#3 Hugging Face Pipeline
## Step 1 – Explain the Concept

A Hugging Face pipeline is a simple API that lets us use a pre-trained model for common AI tasks without writing all the preprocessing and model code ourselves.

Pipeline = Input → Preprocessing → Model → Post-processing → Output

## Text – Hugging Face Pipeline
### Step 1 – Concept

Text pipeline is used to process text using a pre-trained model.

Text → Pipeline → Model → Result

### Step 2 – Simple Example
"I love this product!"
        ↓
Text Pipeline
        ↓
Positive
### Step 3 – Implement

```python
pip install transformers
```

```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")

result = classifier("I love this product")


print(result)
```

| Item               | Meaning                                   |
| ------------------ | ----------------------------------------- |
| **Task**           | Sentiment Analysis                        |
| **Model**          | Pre-trained NLP model                     |
| **Input**          | `"I love this product!"`                  |
| **Processing**     | Tokenize → Model → Prediction             |
| **Output**         | `POSITIVE`                                |
| **Interpretation** | Text has positive sentiment               |
| **Limitation**     | May struggle with sarcasm or unclear text |

## Text Classification
### Step 1 – Explain the Concept

Text Classification means assigning a category/label to a piece of text.

Examples:

Email → Spam / Not Spam
Review → Positive / Negative
News → Sports / Politics / Business
### Step 2 – Simple Example

Input: "I want to cancel my insurance policy."

➡️ Model processes the text
➡️ Output: Insurance Cancellation

### Step 3 – Implement

```python
from transformers import pipeline

classifier = pipeline("text-classification")

result = classifier("i want to cancel my insurance policy")

print(result)

result = classifier("I am very happy with my insurance service.")
print(result)
```

| Part           | Meaning                                      |
| -------------- | -------------------------------------------- |
| Task           | Text Classification                          |
| Model          | Pre-trained NLP model                        |
| Input          | Text                                         |
| Processing     | Text → Tokens → Model                        |
| Output         | Category/Label + Score                       |
| Interpretation | Model predicts the most suitable category    |
| Limitation     | May perform poorly on unclear or unseen text |


## Named Entity Recognition (NER)
### Step 1 – Explain the Concept

NER identifies important entities/names in text and assigns a category to them.

Common entities:

Person
Organization
Location
Date
Money
### Step 2 – Simple Example

Input:
"Mounika works at Cognizant in Hyderabad."

Output:

| Text      | Entity       |
| --------- | ------------ |
| Mounika   | Person       |
| Cognizant | Organization |
| Hyderabad | Location     |


```python
from transformers import pipeline

ner = pipeline("ner", aggregation_strategy="simple")

text = "Mounika works at Cognizant in Hyderabad."

result = ner(text)

print(result)
```

## Text Generation
### Step 1 – Explain the Concept

Text Generation means generating new text based on a given input or prompt using a pre-trained language model.

### Step 2 – Simple Example

Input:
"Artificial Intelligence is"

➡️ Model predicts the next words.

Output:
"Artificial Intelligence is changing the way we work and learn."

### Step 3 – Implement

```python
from transformers import pipeline

generator = pipeline("text-generation")

result = generator(
    "Artificial Intelligence is",
    max_new_tokens=30
)

print(result[0]["generated_text"])
```

“Text generation is an NLP task where a pre-trained language model generates new text based on a given prompt.”

```python
## Image Classification
Image Classification
### Step 1 – Explain the Concept

Image Classification means assigning a class/label to an image.

Example:

Image of apple → Apple
Image of dog → Dog
Image of car → Car
Step 2 – Simple Example
```

```python
## Image-to-Text
### Step 1 – Explain the Concept

Image-to-Text means generating a text description from an image.

It is also called Image Captioning.

Step 2 – Simple Example
```

```python
from transformers import pipeline

captioner = pipeline("image-to-text")

result = captioner("dog_text.jpg")

print(result)
```

```python
### Speech Recognition
Step 1 – Explain the Concept

Speech Recognition means converting spoken audio into text.

It is also called Automatic Speech Recognition (ASR).

Step 2 – Simple Example

🎤 Input:

“Hello, how are you?”

➡️ Speech Recognition Model
➡️ Output:
"Hello, how are you?"

Step 3 – Implement
```

```python
from transformers import pipeline

recognizer = pipeline("automatic-speech-recognition")

result = recognizer("left audio.wav")

print(result["text"])
```
