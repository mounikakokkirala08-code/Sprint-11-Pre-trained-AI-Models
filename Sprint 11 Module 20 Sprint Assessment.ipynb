```python
Pre-trained Models – Important Interview Questions
1. What is a Pre-trained Model?

A pre-trained model is an AI model that has already been trained on a large dataset and can be reused for a specific task.

Example: BERT, ResNet, Whisper.

Pre-trained = Already learned → Reuse for your task

2. What is Inference?

Inference is the process of using a trained model to make predictions on new data.

Input → Trained Model → Prediction

Example:

"I love this product"
        ↓
Model
        ↓
POSITIVE
3. What are Model Weights?

Model weights are the learned numerical values inside a model.

They store the patterns learned during training.

Weights = What the model learned

4. What is a Checkpoint?

A checkpoint is a saved version of a model's weights and training state at a particular point.

Training
  ↓
Checkpoint 1
  ↓
More Training
  ↓
Checkpoint 2

It allows us to resume training or use a saved model.

5. What is a Tokenizer?

A tokenizer converts text into tokens/numbers that a model can understand.

"I love AI"
     ↓
Tokenizer
     ↓
[101, 1045, 2293, 9933, 102]

Tokenizer = Text → Model-readable tokens

6. What is a Model Card?

A Model Card is documentation describing an AI model.

It usually contains:

Model architecture
Intended use
Training data
Input/output
Limitations
Bias considerations
License

Model Card = Model documentation

7. What is Transfer Learning?

Transfer learning means taking knowledge learned by a pre-trained model and applying it to a new related task.

Pre-trained Model
       ↓
Reuse learned knowledge
       ↓
Fine-tune
       ↓
New Task

Example:

ImageNet-trained ResNet → Medical image classification

8. Why Use a Pre-trained Model?

Main reasons:

Saves training time
Requires less data
Reduces computational cost
Often gives better performance
Can be fine-tuned for specific tasks
Comparison Questions
9. Training from Scratch vs Pre-trained Model
Training from Scratch	Pre-trained Model
Starts with random weights	Starts with learned weights
Requires more data	Requires less data
Takes longer	Faster
Higher computational cost	Lower cost
Needs more training	Can fine-tune

Simple:

Scratch = learn from zero
Pre-trained = start with existing knowledge

10. Pipeline vs Manual Model Loading
Pipeline	Manual Loading
Very simple	More control
Less code	More code
Handles preprocessing automatically	You handle preprocessing
Good for beginners/prototyping	Good for customization
Pipeline
from transformers import pipeline

model = pipeline("sentiment-analysis")
Manual
from transformers import AutoTokenizer, AutoModelForSequenceClassification

tokenizer = AutoTokenizer.from_pretrained("model_name")
model = AutoModelForSequenceClassification.from_pretrained("model_name")
11. CPU vs GPU Inference
CPU	GPU
General-purpose processor	Parallel processor
Usually slower for large models	Usually faster
Lower hardware cost	More expensive
Good for small models	Good for large models

CPU → Simple/small workloads
GPU → Large/compute-heavy workloads

12. Small vs Large Model
Small Model	Large Model
Faster inference	Slower inference
Less memory	More memory
Lower cost	Higher cost
Easier deployment	More hardware needed
May have lower performance	May have better performance

Choose based on accuracy requirements + latency + hardware + cost.

13. Accuracy vs Inference Speed

There is often a trade-off:

Larger Model
     ↓
Potentially better accuracy
     ↓
More computation
     ↓
Slower inference
Smaller Model
     ↓
Faster inference
     ↓
Lower cost
     ↓
Potentially lower accuracy
Example
Model	Accuracy	Speed
Small	90%	Fast
Large	95%	Slow

If you need real-time responses, speed may be more important.

If you need maximum accuracy, a larger model may be better.

⭐ Interview Memory
Pre-trained Model → Already trained
Inference         → Make prediction
Weights            → Learned values
Checkpoint         → Saved model state
Tokenizer          → Text → Tokens
Model Card         → Model documentation
Transfer Learning  → Reuse learned knowledge

Model selection rule:

Task + Accuracy + Speed + Size + Hardware + Cost + License + Limitations.
```
